```python
from pyspark import SparkContext
sc = SparkContext.getOrCreate()

# Sample RDD
data = [1, 2, 3, 4, 5]
rdd = sc.parallelize(data)

# Double each number
mapped_rdd = rdd.map(lambda x: x * 2)
print(mapped_rdd.collect())
```

```python
# Sum all elements
total = rdd.reduce(lambda x, y: x + y)
print(total)  # Output: 15
```
