MNIST dataset is a widely used benchmark in the field of machine learning, particularly for training and evaluating neural networks. It consists of a collection of 70,000 labeled images of handwritten digits from 0 to 9. These images are grayscale, 28x28 pixels in size. The dataset is split into 60,000 training images and 10,000 testing images.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
import torchvision.transforms as transforms

# 1. Load data
transform = transforms.ToTensor()
trainset = torchvision.datasets.MNIST(root="./data", train=True, download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=64, shuffle=True)

# 2. Define a simple neural network
class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(28*28, 128)   # input → hidden (Input size: 28×28 = 784 (flattened image), Output: 128 neurons (hidden layer size))
        self.fc2 = nn.Linear(128, 64)      # hidden → hidden (Hidden layer 1: 128 neurons-ReLU activation, Hidden layer 1: 128 neurons-ReLU activation)
        self.fc3 = nn.Linear(64, 10)       # hidden → output (Output layer: 10 neurons (one per digit 0–9))

    def forward(self, x):
        x = x.view(-1, 28*28)              # flatten image (Flattens each 2D image to a 1D vector)
        x = F.relu(self.fc1(x))            # hidden layer 1
        x = F.relu(self.fc2(x))            # hidden layer 2
        return self.fc3(x)                 # raw output (logits)

# 3. Initialize model, loss, optimizer
model = SimpleNN()
criterion = nn.CrossEntropyLoss() # Lower loss = better fit
optimizer = torch.optim.Adam(model.parameters(), lr=0.001) # Adam(model.parameters()=which weights to update, Learning rate (lr) = how big a step to take when updating weights)
# Adam = Adaptive Moment Estimation It’s one of the most popular optimizers in deep learning
# Adam optimizer to adjust model parameters with learning rate 0.001.
# Small lr → slow learning, but stable. Large lr → fast learning, but may overshoot or diverge. 0.001 is a common default for Adam

# 4. Training loop
for epoch in range(5):  # just 2 epochs for demo
    for x, y in trainloader: # x: batch of images (shape: [batch_size (64 images), 1 (in one batch), 28, 28 9of 28 X 28 pixel]) y: batch of labels (shape: [batch_size(64)]).
        optimizer.zero_grad()        # clears gradients from the previous step.
        logits = model(x)            # forward
        loss = criterion(logits, y)  # compute loss
        loss.backward()              # computes gradients via backpropagation
        optimizer.step()             # updates model parameters using Adam

    # So each batch does:
    # 1.Get a small group of images (x) and their labels (y).
    # 2. Pass them through the network.
    # 3. Calculate how wrong the predictions are.
    # 4. Update weights.


    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
```
