```python
from pyspark import SparkContext
sc = SparkContext.getOrCreate()


rdd = sc.parallelize(["hello world", "spark is fast", "map and reduce"])

# Map each sentence to its length
lengths = rdd.map(lambda line: len(line))

# Reduce to total character count
total_chars = lengths.reduce(lambda x, y: x + y)
print(total_chars)  # Total number of characters in all lines
```
